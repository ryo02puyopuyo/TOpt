```python
import numpy as np

# ファイル名（手元のファイル名に合わせて変更してください）
filename = '/home/rest/forpyzx/TOpt/data_for_TOPT/gf27/topt_input_0007.gsm'

try:
    # 行列データを読み込む
    matrix = np.loadtxt(filename)
    
    # 行数と列数の取得
    rows, cols = matrix.shape
    total_elements = matrix.size
    
    # 非ゼロ要素のカウント
    nnz = np.count_nonzero(matrix)
    
    # 密度の計算
    density = nnz / total_elements
    
    print(f"行列のサイズ: {rows}行 x {cols}列")
    print(f"全要素数: {total_elements}")
    print(f"非ゼロ要素数 (1の数): {nnz}")
    print("-" * 30)
    print(f"行列密度: {density:.6f} ({density*100:.2f}%)")

except ValueError as e:
    print("エラー: データの読み込みに失敗しました。")
    print("各行の列数（数字の数）が一致しているか確認してください。")
    print(f"詳細: {e}")
except FileNotFoundError:
    print(f"エラー: ファイル '{filename}' が見つかりませんでした。")
```
